```python
from transformers import BartForConditionalGeneration, BartTokenizer

model_name = "facebook/bart-large-cnn"

tokenizer = BartTokenizer.from_pretrained(model_name)
model = BartForConditionalGeneration.from_pretrained(model_name)
```

```python
sample_article = """Headline: Cyber fraud syndicate busted in ₹12.84 crore Hero FinCorp case, five arrested
Published: September 17, 2026
Source:
The Hindu
Delhi Police arrested five people, including a private-bank relationship manager, in connection with an alleged cyber fraud involving ₹12.84 crore siphoned from the corporate bank account of Hero FinCorp Ltd.
Police said the money was transferred through 317 unauthorised transactions into 34 bank accounts.
Investigators allegedly found that the suspects used “mule accounts”—bank accounts used to receive and move illegally obtained money.
Police also discovered links to Chinese groups operating on Telegram and said part of the money was converted into the cryptocurrency USDT before being transferred abroad.
The investigation led to raids in several states. Police placed a hold, or lien, on approximately ₹1.7 crore in suspected fraudulent accounts and recovered cheque books, mobile phones and rubber stamps.
Authorities said the suspected network involved members in India and Dubai, and that further arrests were expected. The investigation was ongoing at the time of publication.
"""

inputs = tokenizer([sample_article], max_length=1024, return_tensors="pt", truncation=True)

summary_ids = model.generate(inputs["input_ids"], num_beams=4, min_length=30, max_length=130, early_stopping=True)

summary_text = tokenizer.decode(summary_ids[0], skip_special_tokens=True)

print("Original Article Length:", len(sample_article.split()))
print("Summary Length:", len(summary_text.split()))
print("\n--- Generated Summary ---")
print(summary_text)
```
