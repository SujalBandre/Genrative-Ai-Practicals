```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

text = """
deep learning is a powerful technique for artificial intelligence
deep learning is widely used in computer vision
deep learning is useful for natural language processing
deep learning is useful for data science
machine learning is a powerful method for prediction
machine learning is widely used in data science
machine learning is useful for artificial intelligence
natural language processing is useful for text prediction
natural language processing is a branch of artificial intelligence
artificial intelligence is changing the world
artificial intelligence is useful in many applications
lstm networks are useful for sequence prediction
lstm networks are powerful for learning sequential data
lstm networks are widely used for natural language processing
lstm is useful for sequence prediction
sequence prediction is useful for predictive typing
predictive typing is useful for mobile keyboards
predictive typing can suggest the next word
mobile keyboards use language models
language models predict the next word
language models are useful for text generation
text prediction is useful for mobile applications
deep learning models can learn sequential patterns
deep learning models are powerful for natural language processing
"""

tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

total_words = len(tokenizer.word_index) + 1

print("Vocabulary Size:", total_words)


input_sequences = []

for line in text.strip().split("\n"):

    token_list = tokenizer.texts_to_sequences([line])[0]

    for i in range(1, len(token_list)):

        n_gram_sequence = token_list[:i + 1]

        input_sequences.append(n_gram_sequence)

max_sequence_len = max(
    len(seq) for seq in input_sequences
)

input_sequences = np.array(
    pad_sequences(
        input_sequences,
        maxlen=max_sequence_len,
        padding="pre"
    )
)

X = input_sequences[:, :-1]
y = input_sequences[:, -1]

print("Maximum Sequence Length:", max_sequence_len)
print("Training Samples:", len(X))


model = Sequential([
    Embedding(
        input_dim=total_words,
        output_dim=64,
        input_length=max_sequence_len - 1
    ),

    LSTM(128),

    Dense(
        total_words,
        activation="softmax"
    )
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("\nTraining LSTM Model...")

model.fit(X, y, epochs=50, verbose=1)

print("\nModel training completed!")


user_input = input("\nEnter a sentence: ").strip()


token_list = tokenizer.texts_to_sequences([user_input.lower()])[0]

if len(token_list) == 0:
    print("\nSorry, the entered words are not in the dataset.")

else:
    token_list = token_list[-(max_sequence_len - 1):]

    token_list = pad_sequences(
        [token_list],
        maxlen=max_sequence_len - 1,
        padding="pre"
    )


    predictions = model.predict(
        token_list,
        verbose=0
    )[0]

    top_indices = np.argsort(
        predictions
    )[-3:][::-1]

    index_to_word = {
        index: word
        for word, index in tokenizer.word_index.items()
    }

    print("PREDICTION RESULT")

    print("\nInput Sentence:")
    print(user_input)

    print("\nTop-3 Next Word Suggestions:")

    count = 1
    highest_prob_word = ""
    highest_prob_value = -1

    for index in top_indices:

        if index in index_to_word:

            word = index_to_word[index]

            probability = predictions[index] * 100

            print(
                f"{count}. {word} "
                f"({probability:.2f}%)"
            )

            if probability > highest_prob_value:
                highest_prob_value = probability
                highest_prob_word = word

            count += 1

    if highest_prob_word:
        print(f"\nCompleted Sentence: {user_input} {highest_prob_word}")
```
